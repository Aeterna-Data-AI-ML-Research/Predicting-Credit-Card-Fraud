## Method sketch

1. **Consume newer public near-tops** (board-model steps) instead of micro-orthogonalizing the old 0.95426 / 0.95440 manifold.
2. Work in **average-rank → logit** space; map back with average rank-fractions.
3. **BOLD extrapolate** one half-step past F along the F−E axis (same family as earlier dual/extrapolate probes).
4. Nested gates rejected most local natives; the live lift came from **published** board files.

```python
from pathlib import Path
import hashlib
import numpy as np
import pandas as pd
from scipy.special import logit
from scipy.stats import rankdata

EXPECTED_ATH_SHA = "6ea4f9bad68a61753a5c9484b13ec1536c4d5218f56c352e950670a3b078e8b4"
EXPECTED_E_SHA = "f42c7c50720d82ccf651ce5f205e8c9d462a53e7e0e45aa569bec0d138817235"
EXPECTED_F_SHA = "504f06e9b25e989fb0ceca4ed34d04ddbc975eb4b987420009025dcc99fe4413"
INPUT = Path("/kaggle/input")
OUT = Path("/kaggle/working/submission.csv")

def locate(name: str) -> Path:
    hits = sorted(INPUT.rglob(name))
    assert hits, f"missing {name} under {INPUT}"
    return hits[0]

def sha256(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()

def avg_rank_frac(v: np.ndarray) -> np.ndarray:
    v = np.asarray(v, dtype=float)
    return (rankdata(v, method="average") - 0.5) / len(v)

def rank_logit(v: np.ndarray) -> np.ndarray:
    r = np.clip(avg_rank_frac(v), 1e-6, 1.0 - 1e-6)
    return logit(r)

# --- parents from goodpjw Board Model + RealMLP (please upvote that notebook) ---
step_e = locate("goodpjw_step_e_095487.csv")
step_f = locate("goodpjw_step_f_095489.csv")
frozen = locate("submission_095490.csv")
assert sha256(step_e) == EXPECTED_E_SHA, (sha256(step_e), EXPECTED_E_SHA)
assert sha256(step_f) == EXPECTED_F_SHA, (sha256(step_f), EXPECTED_F_SHA)
assert sha256(frozen) == EXPECTED_ATH_SHA, (sha256(frozen), EXPECTED_ATH_SHA)

sample = pd.read_csv(locate("sample_submission.csv"))
E = pd.read_csv(step_e).set_index("id").loc[sample["id"], "is_fraud"].to_numpy()
F = pd.read_csv(step_f).set_index("id").loc[sample["id"], "is_fraud"].to_numpy()

# BOLD F←E @ 0.5  →  Public LB 0.95490 (ref 58521009)
pred = avg_rank_frac(rank_logit(F) + 0.5 * (rank_logit(F) - rank_logit(E)))
sub = pd.DataFrame({"id": sample["id"].to_numpy(), "is_fraud": pred})
assert list(sub.columns) == ["id", "is_fraud"]
assert len(sub) == len(sample)
assert (sub["id"].to_numpy() == sample["id"].to_numpy()).all()
assert sub["is_fraud"].between(0, 1).all()

# Must match the frozen scored file exactly
ref = pd.read_csv(frozen).set_index("id").loc[sample["id"], "is_fraud"].to_numpy()
assert np.allclose(sub["is_fraud"].to_numpy(), ref, atol=0.0, rtol=0.0)

sub.to_csv(OUT, index=False)
print("wrote", OUT)
print("rows", len(sub), "| mean", float(sub["is_fraud"].mean()))
print("sha frozen", EXPECTED_ATH_SHA)
print("credits: goodpjw2008 steps E/F + local BOLD extrapolate @0.5")
sub.head()

```
